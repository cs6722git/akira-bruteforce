* 精確模式：將句子切分成最精確的詞語單位，適合用於需要高準確率的文本分析或自然語言處理任務（**預設模式**）。
* 全模式：掃描句子中**所有可能成詞**的詞語，適合需要快速生成所有可能組合的場合，但無法處理歧義，會產生冗餘的結果。
* 搜索引擎模式：基於精確模式，對長詞進行額外切分以涵蓋更多詞組，適合用於構建搜索引擎索引以提高召回率。

#### 模式選擇

````python
# 1. 精確模式（預設模式）
words_precise = jieba.cut(text, cut_all=False)
print("精確模式分詞   :", " | ".join(words_precise))

# 2. 全模式
words_full = jieba.cut(text, cut_all=True)
print("全模式分詞     :", " | ".join(words_full))

# 3. 搜索引擎模式
words_search = jieba.cut_for_search(text)
print("搜索引擎模式分詞:", " | ".join(words_search))
````

#### 停用詞處理

在分詞後，可以過濾掉一些常見但無意義的停用詞（如「是」、「的」、「和」等）。

````python
# 停用詞列表
stop_words = {"是", "的", "和"}

# 過濾停用詞
words = jieba.cut(text)
# 檢查分詞後的詞是否在停用詞列表中
filtered_words = [word for word in words if word not in stop_words]

print(text)
print("過濾停用詞後:", " | ".join(filtered_words))
````

在自然語言處理中，我們常用 TF-IDF 作為詞語權重的計算方法：
$$
\text{TF-IDF}(t, d) = \text{TF}(t, d) \times \log\frac{N}{\text{DF}(t)}
$$
其中，$N$ 表示文檔總數，$\text{DF}(t)$ 表示包含詞語 $t$ 的文檔數。

#### 使用 TF-IDF 算法提取關鍵字

TF-IDF（Term Frequency-Inverse Document Frequency）是一種常用的關鍵字提取算法，通過計算詞語的重要性，識別文本中的核心內容。該算法的基本原理如下：

- **詞頻 (Term Frequency, TF)**：計算某個詞語在文章中出現的頻率。詞頻越高，該詞語的重要性越可能提升。
- **逆向文件頻率 (Inverse Document Frequency, IDF)**：衡量詞語在所有文件中出現的頻率。詞語在越多文件中出現，其區分性越低，IDF 值越小。

### **公式**
TF-IDF 的計算公式如下：

$$
\text{TF-IDF}(t, d) = \text{TF}(t, d) \times \text{IDF}(t)
$$

- $t$：詞語
- $d$：文件
- $TF(t, d)$：詞語 $t$ 在文件 $d$ 中的詞頻
- $\text{IDF}(t)$：詞語 $t$ 的逆向文件頻率

### **Jieba 提供的 TF-IDF 支持**
- Jieba 自帶了一個默認的 IDF 語料庫，適合通用場景。
- 可以通過 `jieba.analyse.set_idf_path(file_name)` 加載自定義 IDF 文件，以支持專業領域的關鍵字提取。

### **實際應用**
1. 分析文本時，Jieba 根據每個詞的 TF 和 IDF 分數計算 TF-IDF，將詞語的重要性進行排序。
2. 分數越高，該詞語的重要性越大，即更有可能成為關鍵字。

````python
import jieba.analyse

text = """
洛杉磯——日本棒球天才大谷翔平近日正式宣布加盟洛杉磯道奇隊，這位以「投打二刀流」著稱的球員，將成為道奇隊追逐世界大賽冠軍的重要武器。
這一消息引起了棒球界的熱烈討論。大谷翔平過去在天使隊期間創下多項紀錄，包括單賽季打擊和投球數據雙雙進入歷史頂尖行列。
他的驚人表現讓人無法忽視其在球場上的統治力。如今，轉戰道奇隊的大谷，將與球隊現有的明星陣容強強聯手，提升整體競爭力。
「加盟道奇隊是一個令人興奮的決定，這裡有一支極具潛力的球隊和熱情的球迷。我期待著在新球季為大家帶來精彩表現。」大谷在記者會上如是說。
據悉，道奇隊為簽下大谷提供了一份創紀錄的合約，不僅彰顯了球隊的誠意，也反映出大谷翔平在全球棒球界的價值。
球迷們對這次交易充滿期待，認為大谷的加盟將使道奇隊如虎添翼。新賽季即將開啟，大谷翔平與道奇隊能否擦出耀眼火花，值得拭目以待。
"""

# 提取關鍵詞（默認基於 TF-IDF）
# topK：控制返回的關鍵詞數量（預設 20）
# withWeight：是否返回每個關鍵詞的權重（預設 False)
# allowPOS：僅提取指定詞性（如名詞、動詞等）。預設為 None，提取所有詞。
keywords = jieba.analyse.extract_tags(text, topK=5, withWeight=True)

# 輸出關鍵詞及其權重
for keyword, weight in keywords:
    print(f"{keyword}: {weight:.4f}")
````

````python
# 只提取名詞和專有名詞
keywords = jieba.analyse.extract_tags(text,
                                      topK=5,
                                      withWeight=True,
                                      allowPOS=('n', 'nr', 'ns'))
for keyword, weight in keywords:
    print(f"{keyword}: {weight:.4f}")
````

建立自定義詞庫

````python
# 添加自定義詞到 jieba 詞庫
jieba.add_word("大谷翔平")
jieba.add_word("投打二刀流")
jieba.add_word("道奇隊")
jieba.add_word("二刀流")

keywords = jieba.analyse.extract_tags(text, topK=10, withWeight=True)
for keyword, weight in keywords:
    print(f"{keyword}: {weight:.4f}")
````

#### Tokenize：取得開始/結束位置

````python
import jieba

text = "大谷翔平是一位集投打二刀流於一身的日本職業棒球選手，展現了超凡的天賦與實力。"
print(text)

# 精確模式分詞，並返回每個詞的位置
result = jieba.tokenize(text, mode='default')
print("精確模式:")
for token in result:
    print(f"詞語: {token[0]}, 起始位置: {token[1]}, 結束位置: {token[2]}")

# 搜索引擎模式分詞，並返回每個詞的位置
result = jieba.tokenize(text, mode='search')
print("\n搜索引擎模式:")
for token in result:
    print(f"詞語: {token[0]}, 起始位置: {token[1]}, 結束位置: {token[2]}")
````

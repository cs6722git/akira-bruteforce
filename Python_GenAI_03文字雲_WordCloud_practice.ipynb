# PIL(Pillow)
 是 Python 中著名的影像處理套件，以前在 Python2 的時代是 PIL，到 Python3 fork 出一個可相容新版的 Pillow，可以用來轉檔、調色、濾鏡、浮水印甚至創造圖片一堆的功能，雖然不能像 photoshop 之類的軟體一樣強大，但他可以透過程式語言批次處理大量的檔案，在其他需要用到圖片的應用也有很大的幫助。

# from PIL import Image
是一個 Python 程式碼，用於匯入名為 Pillow (先前為 PIL 或 Python Imaging Library) 的套件中的 Image 模組。這個模組提供了處理圖像的各種功能，例如開啟、儲存、縮放、裁切和濾鏡等。

# from wordcloud import WordCloud, ImageColorGenerator
是Python程式碼，它從wordcloud函式庫中匯入兩個重要的類別：WordCloud和ImageColorGenerator。WordCloud是核心類別，用於生成詞雲圖；而ImageColorGenerator則是一個 सहायक類別，它允許詞雲中的單詞顏色模仿一張背景圖片的顏色。
* wordcloud:這是Python的第三方函式庫，專門用於生成詞雲圖（Word Cloud）。是wordcloud函式庫中的核心類別，它提供了生成詞雲圖的基本功能。你可以透過設定參數來控制詞雲的形狀、字體、顏色、背景等。
* ImageColorGenerator: 這是一個輔助類別，用於根據一張圖片的顏色來為詞雲圖中的單詞上色。你需要先載入一張圖片，然後將它傳給ImageColorGenerator來建立一個顏色生成器，再將這個生成器賦值給WordCloud的color_func參數，這樣詞雲的顏色就會與圖片顏色相匹配。

````python
from PIL import Image
import numpy as np

def create_mask_from_image(image_path, threshold=128):
    """
    將彩色圖片轉換為適合作為 WordCloud 的遮罩（mask）。

    參數:
    - image_path: 圖片檔案路徑
    - threshold: 二值化的閾值，預設為 128（亮度中點）

    返回:
    - mask: 遮罩的 NumPy 陣列
    """
    # 1. 載入圖片
    image = Image.open(image_path).resize((IMG_W, IMG_H))

    # 2. 轉換為灰度圖（灰度值範圍 0-255）
    gray_image = image.convert("L")

    # 3. 二值化處理
    binary_image = gray_image.point(lambda x: 255 if x > threshold else 0)

    # 4. 轉換為 NumPy 陣列
    mask = np.array(binary_image)

    return mask

# 圖片轉換成mask
mask = create_mask_from_image("news.jpg", threshold=245)
# 顯示mask
plt.imshow(mask, cmap='gray')
plt.show()
````

````python
my_wordcloud = WordCloud(
    background_color='white',  # 背景顏色設為白色，預設black
    max_words=100,             # 最大顯示的詞數量，限制為 100 個高頻詞
    mask=mask,                 # 遮罩圖案，用於設定文字雲的形狀（需為二值或灰階圖）
    font_path=font,            # 字形檔案的路徑，用於設置字體（確保支持所需語言）
    width=IMG_W,                # 文字雲圖片的寬度
    height=IMG_H,               # 文字雲圖片的高度
    contour_width=2,           # 輪廓線條的寬度
    contour_color='steelblue', # 輪廓線條的顏色，這裡設定為鋼藍色
    stopwords=stopwords,       # 停用詞清單，過濾掉不需要顯示的詞語
).generate(newsword)               # 根據提供的文本（`words`）生成文字雲
````

````python
# 顯示圖片
plt.figure(figsize=(10, 10))
plt.imshow(my_wordcloud)
plt.axis("off")
plt.show()
````
